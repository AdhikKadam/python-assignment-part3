```python
# Import necessary libraries
import requests
from datetime import datetime

# We define the logger early so we can use it throughout the assignment
def log_error(context, error_type, message):
    """Writes a timestamped error entry to error_log.txt."""
    # Get current time formatted as[YYYY-MM-DD HH:MM:SS]
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    log_entry = f"[{timestamp}] ERROR in {context}: {error_type} — {message}\n"
    
    # Append mode ('a') ensures we don't overwrite previous logs
    with open("error_log.txt", "a", encoding="utf-8") as f:
        f.write(log_entry)
```

```python
# --- Part A: Write and Append ---
notes =[
    "Topic 1: Variables store data. Python is dynamically typed.\n",
    "Topic 2: Lists are ordered and mutable.\n",
    "Topic 3: Dictionaries store key-value pairs.\n",
    "Topic 4: Loops automate repetitive tasks.\n",
    "Topic 5: Exception handling prevents crashes.\n"
]

# Write initial notes
with open("python_notes.txt", "w", encoding="utf-8") as file:
    file.writelines(notes)
print("File written successfully.")

# Append additional notes
with open("python_notes.txt", "a", encoding="utf-8") as file:
    file.write("Topic 6: File I/O allows persistent data storage.\n")
    file.write("Topic 7: APIs connect different systems together.\n")
print("Lines appended.\n")


# --- Part B: Read and Search ---
with open("python_notes.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

# Print numbered lines and strip trailing newlines
for i, line in enumerate(lines, start=1):
    print(f"{i}. {line.strip()}")

print(f"\nTotal number of lines: {len(lines)}")

# Ask user for a keyword (Interactive)
keyword = input("\nEnter a keyword to search: ").strip()
print(f"\nSearch results for '{keyword}':")

found = False
for line in lines:
    # Case-insensitive search
    if keyword.lower() in line.lower():
        print(line.strip())
        found = True

if not found:
    print("No matching lines found. Try another keyword.")
```

```python
# Step 1: Fetch and Display Products
def fetch_products():
    url = "https://dummyjson.com/products?limit=20"
    try:
        # Added timeout=5 as per Task 3C
        response = requests.get(url, timeout=5)
        response.raise_for_status() 
        
        data = response.json()
        products = data.get("products",[])
        
        # Print formatted table header
        print(f"{'ID':<4}| {'Title':<30} | {'Category':<15} | {'Price':<8} | {'Rating'}")
        print("-" * 75)
        
        # Print table rows
        for p in products:
            # Truncating title to 30 chars to keep the table clean
            title = p['title'][:30]
            print(f"{p['id']:<4}| {title:<30} | {p['category']:<15} | ${p['price']:<7} | {p['rating']}")
            
        return products

    except requests.exceptions.ConnectionError:
        print("Connection failed. Please check your internet.")
        log_error("fetch_products", "ConnectionError", "No connection could be made")
    except requests.exceptions.Timeout:
        print("Request timed out. Try again later.")
        log_error("fetch_products", "Timeout", "Request took too long")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
        log_error("fetch_products", "Exception", str(e))
        
    return[]

# Execute the function
product_list = fetch_products()
```

```python
# Step 2: Filter and Sort
if product_list:
    # Filter rating >= 4.5
    high_rated =[p for p in product_list if p['rating'] >= 4.5]
    
    # Sort by price descending
    high_rated_sorted = sorted(high_rated, key=lambda x: x['price'], reverse=True)
    
    print("\n--- Filtered (Rating >= 4.5) & Sorted (Price Desc) ---")
    for p in high_rated_sorted:
        print(f"{p['title']} - ${p['price']} (Rating: {p['rating']})")

```

```python
# Step 3: Search by Category (Laptops)
print("\n--- Laptops Category ---")
try:
    response = requests.get("https://dummyjson.com/products/category/laptops", timeout=5)
    laptops = response.json().get("products", [])
    for laptop in laptops:
        print(f"{laptop['title']} - ${laptop['price']}")
except Exception as e:
    print(f"Failed to fetch laptops: {e}")

```

```python
# Part A: Guarded Calculator
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Error: Cannot divide by zero"
    except TypeError:
        return "Error: Invalid input types"

print("--- Guarded Calculator Tests ---")
print(f"10 / 2 = {safe_divide(10, 2)}")
print(f"10 / 0 = {safe_divide(10, 0)}")
print(f"'ten' / 2 = {safe_divide('ten', 2)}")

```

```python

# Part B: Guarded File Reader
def read_file_safe(filename):
    try:
        with open(filename, "r", encoding="utf-8") as f:
            return f.read()
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
    finally:
        print("File operation attempt complete.")

print("\n--- Guarded File Reader Tests ---")
print("Reading 'python_notes.txt':")
content = read_file_safe("python_notes.txt")
if content:
    print("-> File read successfully (content hidden for brevity).")

print("\nReading 'ghost_file.txt':")
read_file_safe("ghost_file.txt")
```

```python
# Part D: Input Validation Loop
print("--- Product Lookup ---")
while True:
    user_input = input("Enter a product ID to look up (1–100), or 'quit' to exit: ").strip()
    
    if user_input.lower() == 'quit':
        print("Exiting lookup.")
        break
        
    # Validate input is an integer
    try:
        product_id = int(user_input)
    except ValueError:
        print("Warning: Invalid input. Please enter a valid integer.")
        continue
        
    # Validate range
    if not (1 <= product_id <= 100):
        print("Warning: Product ID must be between 1 and 100.")
        continue
        
    # Make API Call
    url = f"https://dummyjson.com/products/{product_id}"
    try:
        response = requests.get(url, timeout=5)
        
        if response.status_code == 404:
            print("Product not found.")
        elif response.status_code == 200:
            data = response.json()
            print(f"Found: {data.get('title')} - ${data.get('price')}")
        else:
            print(f"Unexpected status code: {response.status_code}")
            
    except requests.exceptions.ConnectionError:
        print("Connection failed. Please check your internet.")
    except requests.exceptions.Timeout:
        print("Request timed out. Try again later.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
```
